Question 1: Correlation vs. Causation
The map showed a high concentration of deaths near the Broad Street pump. However, proximity alone only proves a correlation. What specific steps did Snow take to move his argument from correlation (deaths happen near the pump) to causation (the pump water is causing the deaths)?

Answer 1:
John Snow proved causation by studying special cases:
People near the pump (like brewery workers and workhouse residents) did not get sick because they did not drink the water.
A woman far from the pump got sick because she did drink the water from it.
This showed that the water from the pump caused the disease, not just being close to it.

Question 2: The Danger of Missing Data
Imagine if Snow had only collected data from the local hospitals, rather than walking the streets and knocking on doors. How would this have changed his dataset, and how might it have altered his conclusions?

Answer 2:
If John Snow had only used hospital data, his findings would have been much weaker. He would not have had exact locations of deaths, so he couldn’t see the clear cluster around the pump. He also would not know where people got their drinking water, which was the key factor. In addition, he would have missed important cases that helped prove his idea. Because of this, he might not have been able to show that the pump water caused cholera or stop the outbreak.

Question 3: The “Miasma” of Today
The “miasma theory” was a widely accepted consensus that was fundamentally incorrect but seemed to explain the surface-level facts (Soho smelled bad, and people in Soho were dying). What is a modern example of a “miasma”—a widely accepted assumption in business, society, or science that might fall apart if rigorously tested with data?

Answer 3:
A modern example of a “miasma” is the belief that open-plan offices increase collaboration. This idea seems logical because removing walls should help people communicate more. However, when tested with data, studies showed the opposite—face-to-face interaction actually decreased, and people relied more on emails and messages. Like the old miasma theory, this assumption sounded correct but was proven wrong when carefully tested with real data.
