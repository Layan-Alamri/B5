Question 1: Correlation vs. Causation
The map showed a high concentration of deaths near the Broad Street pump. However, proximity alone only proves a correlation. What specific steps did Snow take to move his argument from correlation (deaths happen near the pump) to causation (the pump water is causing the deaths)?

Answer 1:
John Snow proved causation by studying special cases:
People near the pump (like brewery workers and workhouse residents) did not get sick because they did not drink the water.
A woman far from the pump got sick because she did drink the water from it.
This showed that the water from the pump caused the disease, not just being close to it.
