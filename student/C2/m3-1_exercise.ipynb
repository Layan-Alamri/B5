## B. Calculations on the sample: proportion, mean, variance

Use the cleaned numeric columns only where needed (drop rows with missing body mass or bill length, or use `.dropna(subset=[...])` as you prefer). Show your code and round printed numbers sensibly.

**1.** For the variable `body_mass_g`:

- Compute the **proportion** of penguins with body mass **greater than** the overall median body mass (among non-missing values).
- Compute the **mean** and **variance** of `body_mass_g` (same subset).

```python
# YOUR CODE
df_clean = df.dropna(subset=["body_mass_g"])

median_mass = df_clean["body_mass_g"].median()
prop = (df_clean["body_mass_g"] > median_mass).mean()

mean_mass = df_clean["body_mass_g"].mean()
var_mass = df_clean["body_mass_g"].var()

print("Median:", round(median_mass, 2))
print("Proportion > median:", round(prop, 3))
print("Mean:", round(mean_mass, 2))
print("Variance:", round(var_mass, 2))
```

**2.** For `body_mass_g`, identify at least **two** distributional summary measures that describe **center** and **spread** (name them and compute them). Optionally add a third that describes **shape** (e.g. skewness), if you know how to compute it.

```python
# YOUR CODE
mean_mass = df_clean["body_mass_g"].mean()
median_mass = df_clean["body_mass_g"].median()

std_mass = df_clean["body_mass_g"].std()
iqr_mass = df_clean["body_mass_g"].quantile(0.75) - df_clean["body_mass_g"].quantile(0.25)

skew_mass = df_clean["body_mass_g"].skew()

print("Mean:", round(mean_mass, 2))
print("Median:", round(median_mass, 2))
print("Standard Deviation:", round(std_mass, 2))
print("IQR:", round(iqr_mass, 2))
print("Skewness:", round(skew_mass, 3))
```

Work on `body_mass_g` (non-missing) or `bill_length_mm`—pick one and stick to it for this section.

**3.** For a normal distribution, explain what share of values fall within **1**, **2**, and **3 standard deviations** of the mean (the empirical rule). Then **compute** the interval $\mu \pm 1\sigma$, $\mu \pm 2\sigma$, and $\mu \pm 3\sigma$ using your sample $\mu$ and $\sigma$ for that variable.

For a normal distribution:

About 68% of values fall within 1 standard deviation of the mean
About 95% of values fall within 2 standard deviations of the mean
About 99.7% of values fall within 3 standard deviations of the mean

```python
# YOUR CODE
df_flipper = df.dropna(subset=["flipper_length_mm"])

mu = df_flipper["flipper_length_mm"].mean()
sigma = df_flipper["flipper_length_mm"].std()

interval_1 = (mu - sigma, mu + sigma)
interval_2 = (mu - 2*sigma, mu + 2*sigma)
interval_3 = (mu - 3*sigma, mu + 3*sigma)

print("Mean:", round(mu, 2))
print("Standard deviation:", round(sigma, 2))

print("μ ± 1σ:", (round(interval_1[0], 2), round(interval_1[1], 2)))
print("μ ± 2σ:", (round(interval_2[0], 2), round(interval_2[1], 2)))
print("μ ± 3σ:", (round(interval_3[0], 2), round(interval_3[1], 2)))
```
